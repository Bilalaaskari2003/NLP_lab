Task7

```python
# (a)
text = "NLP is the future of AI research"
print("Starts with 'NLP':", text.startswith("NLP"))
print("Ends with 'research':", text.endswith("research"))
print("Position of 'future':", text.find("future"))

# (b)
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        label = "purely alphabetic"
    elif token.isdigit():
        label = "purely numeric"
    else:
        label = "mixed"
    print(f"{token}: {label}")

# (c)
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("pure_words:", pure_words)
print("pure_numbers:", pure_numbers)
print("mixed_tokens:", mixed_tokens)

# (d)
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

for token in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(token, "->", validate_token(token))
```

Task8

```python
# (a)
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

test_text = " HELLO Natural Language Processing "
print("clean():", clean(test_text))
print("clean_lambda():", clean_lambda(test_text))

# (b)
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]

sentence = "I love working with NLP tools in Python daily"
result = remove_short(tokenize(clean(sentence)), 3)
print("remove_short result:", result)

# (c)
def word_lengths(tokens):
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print("word_lengths:", word_lengths(tokens))

# (d)
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

print("Pipeline 1:", nlp_pipeline("I love working with NLP tools in Python daily", 3))
print("Pipeline 2:", nlp_pipeline("Deep learning models process natural language text", 4))
```

```python
# (a)
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lower_words = list(map(lambda w: w.lower(), words))
word_lengths_list = list(map(lambda w: len(w), words))

print("Lowercase words:", lower_words)
print("Character lengths:", word_lengths_list)

# (b)
words_long = list(filter(lambda w: len(w) > 3, words))
words_with_t = list(filter(lambda w: "t" in w.lower(), words))

print("words_long:", words_long)
print("words_with_t:", words_with_t)

# (c)
print("Alphabetical:", sorted(words))
print("By length ascending:", sorted(words, key=len))
print("By length descending:", sorted(words, key=len, reverse=True))

# (d)
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
split_words = sentence.split()
freq = Counter(split_words)

print("Full Counter:", freq)
print("3 most common words:", freq.most_common(3))
print("Frequency of 'python':", freq["python"])
print("Total number of words:", sum(freq.values()))
print("Unique vocabulary size:", len(freq))
```
